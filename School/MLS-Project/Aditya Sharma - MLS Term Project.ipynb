```python
import os
import logging
import zipfile
import pandas as pd
```

# **1. Loading Datasets**

```python
path_to_datasets = {0: r'datasets\dataset_1',
										1: r'datasets\dataset_2',
										2: r'datasets\dataset_3'}
dataset_zip = {0: r'datasets\dataset_1.zip',
						   1: r'datasets\dataset_2.zip',
							 2: r'datasets\dataset_3.zip'}
expected_files = {0: [os.path.join(path_to_datasets[0],
																	 file) for file in [r'fashion-mnist_train.csv',
																		 									r'fashion-mnist_test.csv',
																											r'train-images-idx3-ubyte',
																											r'train-labels-idx1-ubyte',
																											r't10k-images-idx3-ubyte',
																											r't10k-labels-idx1-ubyte']],
								  1: [os.path.join(path_to_datasets[1], r'vehicles.csv')],
								  2: [os.path.join(path_to_datasets[2], r'TSLA.csv')]}

logging.basicConfig(level=logging.INFO,
										format='%(asctime)s - %(levelname)s - %(message)s')

def ensure_directory_exists(directory_path):
	if not os.path.exists(directory_path):
		os.makedirs(directory_path,
							  exist_ok=True)
		logging.info(f"Created directory: {directory_path}")
	else:
		logging.info(f"Directory exists: {directory_path}")
def extract_zip(zip_path,
								extract_to,
								expected_files):
	if os.path.exists(zip_path):
		logging.info(f"Zip file exists: {zip_path}")

		if all(os.path.exists(file) for file in expected_files):
			logging.info(f"All expected files already extracted in: {extract_to}")
		else:
			try:
				with zipfile.ZipFile(zip_path,
												 		 'r') as zip_ref:
					zip_contents = zip_ref.namelist()
					missing_files = [file for file in expected_files if os.path.basename(file) not in zip_contents]

					if missing_files:
						logging.error(f"Missing files in ZIP: {missing_files}")

						return
					
					logging.info(f"Extracting files from {zip_path}...")

					zip_ref.extractall(extract_to)

					logging.info(f"Files extracted to: {extract_to}")
			except zipfile.BadZipFile:
				logging.error(f"Failed to extract {zip_path}: Bad ZIP file.")
	else:
		logging.warning(f"Zip file not found: {zip_path}. Skipping extraction.")

for [key,
		 dataset_folder] in path_to_datasets.items():
	zip_file = dataset_zip[key]
	expected = expected_files[key]

	ensure_directory_exists(dataset_folder)
	extract_zip(zip_file,
						  dataset_folder,
							expected)
```
