```python
# import relevant Python libraries

import os
import zipfile
```

# **1. Loading Datasets**

```python
path_to_datasets = {}
dataset_zip = {}
path_to_datasets[0] = r'C:\Users\Aditya Sharma\Downloads\Projects\School\MLS-Project\datasets\dataset_1'
path_to_datasets[1] = r'C:\Users\Aditya Sharma\Downloads\Projects\School\MLS-Project\datasets\dataset_2'
path_to_datasets[2] = r'C:\Users\Aditya Sharma\Downloads\Projects\School\MLS-Project\datasets\dataset_3'
dataset_zip[0] = 'dataset_1.zip'
dataset_zip[1] = 'dataset_2.zip'
dataset_zip[2] = 'dataset_3.zip'

for i in range(len(path_to_datasets)):
	if not os.path.exists(path_to_datasets[i]):
		os.makedirs(path_to_datasets[i],
							exist_ok=True)

		print(f"Created directory: {path_to_datasets[i]}")
	else:
		print(f"Directory exists: {path_to_datasets[i]}")

	local_filename = os.path.join(path_to_datasets[i],
															 dataset_zip[i])

	if os.path.exists(local_filename):
		print(f"Zip file exists: {local_filename}")
		print(f"Extracting files from {local_filename}...")

		with zipfile.ZipFile(local_filename,
											 'r') as zip_ref:
			zip_ref.extractall(path_to_datasets[i])

		print(f"Files extracted to: {path_to_datasets[i]}")
	else:
		print(f"Zip file not found: {local_filename}. Skipping extraction.")
```
