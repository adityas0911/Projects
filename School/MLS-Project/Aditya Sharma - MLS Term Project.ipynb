# **1. Loading Datasets**

## 1.1

```python
path_to_datasets = r'C:\Users\Aditya Sharma\Downloads\Projects\School\MLS-Project\datasets'

os.makedirs(path_to_datasets,
						exist_ok=True)

url_of_dataset_1 = 'https://www.kaggle.com/datasets/zalando-research/fashionmnist/download?datasetVersionNumber=4'
local_filename = os.path.join(path_to_datasets,
															'fashionmnist.zip')

print("Downloading dataset...")

with requests.get(url_of_dataset_1,
									stream=True) as response:
	response.raise_for_status()
	with open(local_filename,
					 'wb') as file:
		for chunk in response.iter_content(chunk_size=8192):
			file.write(chunk)

print("Dataset downloaded to:",
			local_filename)
print("Extracting files...")

with zipfile.ZipFile(local_filename,
										 'r') as zip_ref:
	zip_ref.extractall(path_to_datasets)

print("Files extracted to:",
			path_to_datasets)

os.remove(local_filename)

print("Deleted ZIP file:",
			local_filename)
```
