```python
# import relevant Python libraries

import os
import zipfile
```

# **1. Loading Datasets**

```python
path_to_datasets = {0: r'datasets\dataset_1',
                    1: r'datasets\dataset_2',
                    2: r'datasets\dataset_3'}
dataset_zip = {0: r'datasets\dataset_1.zip',
               1: r'datasets\dataset_2.zip',
               2: r'datasets\dataset_3.zip'}
expected_files = {0: [r'datasets\dataset_1\fashion-mnist_train.csv',
                      r'datasets\dataset_1\fashion-mnist_test.csv',
                      r'datasets\dataset_1\train-images-idx3-ubyte',
                      r'datasets\dataset_1\train-labels-idx1-ubyte',
                      r'datasets\dataset_1\t10k-images-idx3-ubyte',
                      r'datasets\dataset_1\t10k-labels-idx1-ubyte'],
                  1: [r'datasets\dataset_2\vehicles.csv'],
                  2: [r'datasets\dataset_3\TSLA.csv']}

for i in range(len(path_to_datasets)):
	dataset_folder = path_to_datasets[i]
	zip_file = dataset_zip[i]

	if not os.path.exists(dataset_folder):
		os.makedirs(dataset_folder,
							  exist_ok=True)

		print(f"Created directory: {dataset_folder}")
	else:
		print(f"Directory exists: {dataset_folder}")

	local_filename = os.path.join(dataset_folder,
															  zip_file)
	
	if os.path.exists(local_filename):
		print(f"Zip file exists: {local_filename}")

		all_files_exist = all(os.path.exists(os.path.join(dataset_folder,
																										  file)) for file in expected_files[i])

		if all_files_exist:
			print(f"All expected files already extracted in: {dataset_folder}")
		else:
			print(f"Extracting files from {local_filename}...")

			with zipfile.ZipFile(local_filename,
												   'r') as zip_ref:
				zip_ref.extractall(dataset_folder)

			print(f"Files extracted to: {dataset_folder}")
	else:
		print(f"Zip file not found: {local_filename}.")
```
